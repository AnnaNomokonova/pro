```python
import pandas as pd

# Загружаем данные
df = pd.read_csv('../data/SampleSuperstore.csv')

# 1. Удаляем дубликаты
df = df.drop_duplicates()

# 2. Проверяем пропуски (пустые значения)
print("Пропуски в данных:\n", df.isnull().sum())

# 3. Сохраняем очищенные данные
df.to_csv('../data/cleaned_data.csv', index=False)
print("✅ Данные очищены и сохранены в cleaned_data.csv!")
```
